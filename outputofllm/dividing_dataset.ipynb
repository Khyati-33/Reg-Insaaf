```python
import pandas as pd
import zipfile
from google.colab import files

# Step 1: Load dataset
df = pd.read_csv("/content/drive/MyDrive/lab 4 cv/2500-total.csv")

# Step 2: (Optional) Shuffle dataset
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Step 3: Split into 5 parts (500 rows each)
file_names = []

for i in range(5):
    start = i * 500
    end = (i + 1) * 500

    part = df.iloc[start:end]
    file_name = f"dataset_part_{i+1}.csv"

    part.to_csv(file_name, index=False)
    file_names.append(file_name)

print("✅ Dataset split into 5 parts!")

# Step 4: Zip all files
zip_filename = "dataset_parts.zip"

with zipfile.ZipFile(zip_filename, 'w') as zipf:
    for file in file_names:
        zipf.write(file)

print("✅ Zip file created!")

# Step 5: Download zip file
files.download(zip_filename)
```
